```python
# FastAPI pytest Setup

This notebook inspects the repository, creates a FastAPI app and pytest tests, updates `requirements.txt`, and runs the tests.
</VSCode.Cell>
<VSCode.Cell language="python">
# Install and import dependencies
!pip install fastapi pytest httpx

import os
from pathlib import Path

root = Path("/workspaces/skills-getting-started-with-github-copilot")
print("Repo root:", root)
print("Files in root:", [p.name for p in root.iterdir() if p.is_file()])
</VSCode.Cell>
<VSCode.Cell language="markdown">
## Define FastAPI App

Create a minimal FastAPI application in `src/api.py` if it does not already exist.
</VSCode.Cell>
<VSCode.Cell language="python">
app_path = root / "src" / "api.py"
app_path.parent.mkdir(parents=True, exist_ok=True)
app_code = '''from fastapi import FastAPI

app = FastAPI()

@app.get("/")
async def read_root():
    return {"message": "Hello from FastAPI"}

@app.get("/items/{item_id}")
async def read_item(item_id: int, q: str | None = None):
    return {"item_id": item_id, "q": q}
'''
app_path.write_text(app_code)
print(f"Written FastAPI app to {app_path}")
</VSCode.Cell>
<VSCode.Cell language="markdown">
## Create Tests Directory and Test Files

Create a new `tests` directory and add pytest test cases for the FastAPI endpoints.
</VSCode.Cell>
<VSCode.Cell language="python">
tests_path = root / "tests"
tests_path.mkdir(exist_ok=True)
test_file = tests_path / "test_api.py"
test_code = '''from fastapi.testclient import TestClient

from src.api import app

client = TestClient(app)


def test_read_root():
    response = client.get("/")
    assert response.status_code == 200
    assert response.json() == {"message": "Hello from FastAPI"}


def test_read_item_without_query():
    response = client.get("/items/42")
    assert response.status_code == 200
    assert response.json() == {"item_id": 42, "q": None}


def test_read_item_with_query():
    response = client.get("/items/42?q=example")
    assert response.status_code == 200
    assert response.json() == {"item_id": 42, "q": "example"}
'''
test_file.write_text(test_code)
print(f"Written tests to {test_file}")
</VSCode.Cell>
<VSCode.Cell language="markdown">
## Update requirements.txt

Add FastAPI, pytest, and httpx to the root `requirements.txt` file.
</VSCode.Cell>
<VSCode.Cell language="python">
req_path = root / "requirements.txt"
if req_path.exists():
    current = req_path.read_text().strip().splitlines()
    packages = {line.strip() for line in current if line.strip()}
    for pkg in ["fastapi", "pytest", "httpx"]:
        if pkg not in packages:
            packages.add(pkg)
    req_path.write_text("\n".join(sorted(packages)) + "\n")
    print(f"Updated {req_path} with: fastapi, pytest, httpx")
else:
    req_path.write_text("fastapi\npytest\nhttpx\n")
    print(f"Created {req_path}")
</VSCode.Cell>
<VSCode.Cell language="markdown">
## Run Tests with pytest

Execute pytest in the project root to verify the FastAPI endpoints.
</VSCode.Cell>
<VSCode.Cell language="python">
!cd /workspaces/skills-getting-started-with-github-copilot && pytest -q
</VSCode.Cell>
```
